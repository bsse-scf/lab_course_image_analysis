# Exercise: Python Basics

Short practice tasks to reinforce Python fundamentals: variables, data types, control flow, and functions.

1. Define variables that hold image metadata and print them (e.g. filename, width, height, and pixel_size).

1. Given a pixel intensity value, print if it is above or below a threshold.

1. Simulate iterating over a list of pixel intensities and count how many are above a threshold.

1. Write a function `average_intensity` that takes a list of pixel intensities and returns the average intensity.

1. Once you defined the function, execute the following in a cell: `average_intensity?` to see the function's docstring and signature. Add a docstring to your function to describe what it does.

1. Using the same `?` trick, check the docstring of `tifffile.imread` to see what arguments it takes and what it returns.
